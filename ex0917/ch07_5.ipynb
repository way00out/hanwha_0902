```python
from langchain_core.callbacks import get_usage_metadata_callback
from langchain_openai import ChatOpenAI

llm = ChatOpenAI(model="gpt-5-mini")
```

```python
with get_usage_metadata_callback() as cb: # get_usage_metadata_callback(): 실행되는 모든 토큰 사용량/요금 추적 func
    result = llm.invoke("대한민국의 수도는 어디야?")
    print(cb)
```

```python
with get_usage_metadata_callback() as cb:
    result = llm.invoke("대한민국의 수도는 어디야?")
    result = llm.invoke("대한민국의 수도는 어디야?")
    usage = cb.usage_metadata["gpt-5-mini-2025-08-07"]
    print(f"총 사용된 토큰수: \t\t{usage['total_tokens']}")
    print(f"프롬프트에 사용된 토큰수: \t{usage['input_tokens']}")
    print(f"답변에 사용된 토큰수: \t{usage['output_tokens']}")
```
