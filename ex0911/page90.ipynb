```python
from langchain_openai import ChatOpenAI
```

```python
llm = ChatOpenAI(
    temperature = 0.1,
    model_name = "gpt-5-mini",
)
```

```python
question = "Who is Korea president?"

print(f"[ask]: {llm.invoke(question)}")
```

```python
question ="Who is South Korea president?"

response = llm.invoke(question)
response
```

```python
response.content
```

```python
response.response_metadata
```

```python
response.response_metadata["token_usage"]
```

```python
llm_with_logprob = ChatOpenAI(
    temperature=0.1,
    max_tokens=2048,
    model_name="gpt-4o-mini",
).bind(logprobs=True)
```

```python
question = "Who is South Korea president?"

response = llm_with_logprob.invoke(question)
response.response_metadata
```

```python
answer = llm.stream("빨간색과 어울리는 색 조합 3가지 알려줘")

for token in answer:
    print(token.content, end="", flush=True)
```

```python
answer = llm.stream("빨간색과 어울리는 3가지 색 조합만 알려줘")

final_answer = ""
for token in answer:
    final_answer += token.content
```

```python
print(final_answer)
```
