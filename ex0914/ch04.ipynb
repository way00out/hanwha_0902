```python
from dotenv import load_dotenv

load_dotenv()
```

```python
from langchain_openai import ChatOpenAI
```

```python
llm = ChatOpenAI(
    temperature=0.1,
    model_name="gpt-4o-mini",
)

question = "대한민국의 수도는 어디인가요?"

response = llm.invoke(question)
```

```python
print(f"[답변]: {response}")
```

```python
response.content
```

```python
response.response_metadata
```

```python
response.response_metadata["token_usage"]["total_tokens"]
```

```python
llm_with_logprob = ChatOpenAI(
    temperature=0.1,
    max_tokens=2048,
    model_name="gpt-4o-mini",
).bind(logprobs=True)
```

```python
question = "캐나다의 수도는 어디인가요?"

response = llm_with_logprob.invoke(question)
```

```python
response.content
```

```python
response.response_metadata["logprobs"]
```

```python
answer = llm.stream("커피에 어울리는 디저트 3개 알려줘")

for token in answer:
    print(token.content, end="", flush=True)
```

```python
answer = llm.stream("커피와 어울리는 디저트 3개 알려줘")

final_answer = ""
for token in answer:
    final_answer += token.content
```

```python
print(final_answer)
```
